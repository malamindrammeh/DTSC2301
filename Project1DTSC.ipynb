```python
%pip install lxml

import pandas as pd

# BBC Sport - simple and static-ish HTML
url = "https://www.bbc.co.uk/sport/football/premier-league/table"

tables = pd.read_html(url)
df = tables[0]  # Usually the first table is the PL table

print(df)
```

```python
df['Form'] = df['Form'].str.replace('DResult Draw', 'D')
df['Form'] = df['Form'].str.replace('WResult Win', 'W')
df['Form'] = df['Form'].str.replace('LResult Loss', 'L')
display(df['Form'])
```

```python
df['Team'] = df['Team'].str.replace('\d+', '', regex=True).str.strip()
display(df)
```

```python
df.to_csv('premier_league_table_16.csv', index=False)
```

```python
df['Pyth'] = df['Goals For']**2 /(df['Goals For']**2 + df['Goals Against']**2)
display(df)
```

```python
import numpy as np

# simulate distributions of added points over 12 games per team, then boxplot ordered by median total points
np.random.seed(42)

N = 1000  # number of simulations
games = 12
teams = df['Team'].tolist()
base_points = df['Points'].values
recent = df['Recent_PPG'].values

sim_totals = []
medians = []

for i, team in enumerate(teams):
    # simulate points per game for 12 games with some variability, clip to [0,3]
    sim_ppg = np.random.normal(loc=recent[i], scale=0.7, size=(N, games))
    sim_ppg = np.clip(sim_ppg, 0, 3)
    added = sim_ppg.sum(axis=1)  # total added points over 12 games
    total = base_points[i] + added
    sim_totals.append(total)
    medians.append(np.median(total))

# order teams by median total points (descending)
order = np.argsort(medians)[::-1]
ordered_totals = [sim_totals[i] for i in order]
ordered_names = [teams[i] for i in order]

# plot
plt.figure(figsize=(16,7))
plt.boxplot(ordered_totals, labels=ordered_names, showfliers=False)
plt.xticks(rotation=45, ha='right')
plt.ylabel('Total Points after 12 simulated games')
plt.title('Projected Total Points Distribution (12 more games) — teams ordered by median (high to low)')
plt.tight_layout()
plt.show()
```
